This notebook trains unsupervised anomaly detection models (Isolation Forest) on multidimensional DCGM telemetry signals. Unsupervised methods flag subtle degradation (e.g. stragglers, novel cooling failure modes) without requiring explicit historical labels.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.ensemble import IsolationForest
from sklearn.preprocessing import StandardScaler

df = pd.read_csv("../data/sample_telemetry.csv")
features = ["dcgm_gpu_temp", "dcgm_power_usage", "dcgm_gpu_utilization", "performance_ratio"]
X = df[features]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

iso = IsolationForest(contamination=0.10, random_state=42)
df["anomaly_score"] = -iso.fit_predict(X_scaled) # 1 = anomaly, -1 = normal

print("Anomaly detection complete.")
print(df.groupby("failure_type")["anomaly_score"].mean())
```
